```python
# funkcja podnosząca liczbę do potęgi naturalnej
def power(base, exponent):
    result = 1
    for _ in range(exponent):
        result *= base
    return result
```

```python
# funkcja licząca moduł
def f(x):
    if isinstance(x, complex):
        return (power(x.real,2) + power(x.imag,2)) ** 0.5
    elif x >= 0:
        return x
    else:
        return -x
```

```python
# zdefiniowanie ciągu an
def a_n(n):
    return power(-1/2,n)
```

```python
# zdefiniowanie ciągu bn
def b_n(n):
    return power(1j/3,n)
```

```python
def in_hilbert_space(func, tol=1e-6, max_iter=10**6):
    total = 0.0
    prev_total = 0.0
    n = 1
    reason = ""

    while n <= max_iter:
        prev_total = total
        total += power(f(func(n)), 2)

        # warunek stopu: różnica między kolejnymi sumami bardzo mała
        if abs(total - prev_total) < tol:
            reason = f"Zbieżność osiągnięta (różnica < {tol}) po {n} iteracjach."
            break

        n += 1

    if n > max_iter:
        reason = f"Osiągnięto maksymalną liczbę iteracji ({max_iter}), brak pełnej zbieżności."

    return total, n, reason
```

```python
sum_a, iter_a, reason_a = in_hilbert_space(a_n)

print("Suma dla a_n =", sum_a)
print(reason_a)
```

```python
sum_b, iter_b, reason_b = in_hilbert_space(b_n)

print("Suma dla b_n =", sum_b)
print(reason_b)
```

Oba ciągi należą do przestrzeni Hilberta $l^2$, ponieważ 
$$
\sum_{n=1}^{\infty} |a_n|^2 < \infty \text{ i } \sum_{n=1}^{\infty} |b_n|^2 < \infty.
$$
Z tego wynika, że oba ciągi także należą do wszystkich przestrzeni Banacha $l^p$ dla $p\geq 2$.

```python
#funkcja licząca normę
def norm(func):
    return func ** 0.5
```

```python
def inner_product(func1, func2, tol=1e-6, max_iter=10**6):
    total = 0.0
    prev_total = 0.0
    n = 1
    reason = ""

    while n <= max_iter:
        prev_total = total
        total += func1(n) * func2(n).conjugate()

        if abs(total - prev_total) < tol:
            reason = f"Zbieżność osiągnięta (różnica < {tol}) po {n} iteracjach."
            break

        n += 1

    if n > max_iter:
        reason = f"Osiągnięto maksymalną liczbę iteracji ({max_iter}), brak pełnej zbieżności."

    return total, n, reason
```

```python
ip, iter_ip, reason_ip = inner_product(a_n, b_n)
print("<a, b> =", ip)
print(reason_ip)
```
