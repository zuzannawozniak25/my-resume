```python
# import pakietów
from scipy import integrate
import sympy as sp
from sympy.abc import t, omega
```

Postać nieuniatrna ciągłoczasowego przekształcenia Fouriera sygnału

```python
#wymuszenie korzystania z wartości rzeczywistej
t, omega = sp.symbols('t omega', real=True)
```

```python
# definiujemy funkcję x(t)
x_t = sp.exp(-2*t**2 + 1)
```

```python
x_t
```

```python
# Nieunitarne przekształcenie Fouriera
X_omega = sp.integrate(x_t * sp.exp(-sp.I * omega * t), (t, -sp.oo, sp.oo))
```

```python
X_omega.simplify()
```

```python
# Przeksztalcenie odwrotne
X_inv = 1/(2*sp.pi)*sp.integrate(X_omega* sp.exp(sp.I * omega * t), (omega, -sp.oo, sp.oo))
```

```python
X_inv.simplify()
```

Sygnał został poprawnie odtworzony.

```python
# Wzór na widmową gęstość energii
PS_omega = abs(X_omega)**2
```

```python
PS_omega.simplify()
```

```python
# Twierdzenie zachowania energii
Left = sp.integrate(abs(x_t)**2, (t, -sp.oo, sp.oo))
Right = 1/(2*sp.pi)*sp.integrate((abs(X_omega))**2, (omega, -sp.oo, sp.oo))
```

```python
# Wyświetlenie lewej strony równania
Left
```

```python
# Wyświetlenie prawej strony równania
Right
```

```python
# Sprawdzenie czy obie strony się zgadzają
Left == Right
```

Zachodzi twierdzenie o zachowaniu energii, obie strony równania dały ten sam poprawny rezultat.
