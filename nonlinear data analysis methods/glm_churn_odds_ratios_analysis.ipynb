```python
#importowanie bibliotek
import pandas as pd
import numpy as np
import sympy as sp
from sympy.abc import x,y 
```

```python
#wczytanie danych
df = pd.read_csv("dane.txt")
```

```python
#usunięcie kropek w zmiennej Churn?
df['Churn?'] = df['Churn?'].replace({'True.': True, 'False.': False})
```

```python
#utworzenie nowej ramki danych df2 ze zmiennymi Churn? oraz CustServ Calls
df2 = df[['Churn?', 'CustServ Calls']].copy()
```

```python
df2
```

```python
#utworzenie nowej zmiennej CSC

def csc_kategoria(x):
    if x == 0 or x == 1:
        return 'male'
    elif x == 2 or x == 3 :
        return 'srednie'
    else:
        return 'wysokie'

df2.loc[:, 'CSC'] = df2['CustServ Calls'].apply(csc_kategoria)
```

```python
df2
```

```python
#utworzeie tabeli krzyżowej 
tabk = pd.crosstab(df2['Churn?'], df2['CSC'], margins = True)
```

```python
tabk
```

```python
#nowe zmienne wskaźnikowe
def csc_sr(x):
    if x == 2 or x == 3:
        return 1
    else:
        return 0

def csc_wys(x):
    if x >= 4:
        return 1
    else:
        return 0

df2.loc[:, 'CSC_sr'] = df2['CustServ Calls'].apply(csc_sr)
df2.loc[:, 'CSC_wys'] = df2['CustServ Calls'].apply(csc_wys)
```

```python
df2.head(15)
```

```python
male_true = tabk.loc[True, 'male']
male_false = tabk.loc[False, 'male']
```

```python
OR_nis = male_true/male_false
print("Iloraz liczby klientów, którzy odbyli małą liczbę rozmów i przenieśli się do innego operatora przez liczbę klientów,\nktórzy odbyli małą liczbę rozmów i nie przenieśli się wynosi: ", OR_nis)
```

```python
srednie_true = tabk.loc[True, 'srednie']
srednie_false = tabk.loc[False, 'srednie']
```

```python
OR_sr = (srednie_true * male_false) / (male_true * srednie_false)
print("Iloraz iloczynu liczby klientów, którzy odbyli średnią liczbę rozmów i przenieśli się oraz liczby klientów, \nktórzy odbyli mało rozmów i nie przenieśli się przez iloczyn liczby klientów, którzy odbyli mało rozmów i przenieśli się oraz liczby klientów, \nktórzy odbyli średnią liczbę rozmów i nie przenieśli się wynosi: ", OR_sr)
```

```python
wysokie_true = tabk.loc[True, 'wysokie']
wysokie_false = tabk.loc[False, 'wysokie']
```

```python
OR_wys = (wysokie_true * male_false) / (male_true * wysokie_false)
print("Iloraz iloczynu liczby klientów, którzy odbyli wysoką liczbę rozmów i przenieśli się oraz liczby klientów, \nktórzy odbyli mało rozmów i nie przenieśli się przez iloczyn liczby klientów, którzy odbyli mało rozmów i przenieśli się oraz liczby klientów, \nktórzy odbyli wysoką liczbę rozmów i nie przenieśli się wynosi: ", OR_wys)
```

```python
#obliczanie b0, b1 i b2
b0 = np.log(OR_nis)
b1 = np.log(OR_sr)
b2 = np.log(OR_wys)

print("b0 = ln(OR_nis) wynosi: ", b0)
print("b1 = ln(OR_sr) wynosi: ", b1)
print("b2 = ln(OR_wys) wynosi: ", b2)
```

```python
#zapisanie funkcji logitowej 
x,y = sp.symbols('x y')
g = b0 + b1 * x + b2 * y
print("Funkcja logitowa g(X) = b0 + b1x + b2y, gdzie X=[x,y] ma postać: ", g)
```

```python
#prawdopodobieństwo przeniesienia 
m = sp.exp(b0 + b1 * x + b2 * y) / (1 + sp.exp(b0 + b1 * x + b2 * y))
print("Prawdopodobieństwo przeniesienia się wynosi: ")
m
```

```python
#obliczenie prawdopodobienstwa
p_male = sp.N(sp.exp(b0)/(1 + sp.exp(b0)))
p_srednie = sp.N(sp.exp(b0 + b1)/(1 + sp.exp(b0 + b1)))
p_wysokie = sp.N(sp.exp(b0 + b2)/(1 + sp.exp(b0 + b2)))

print("Prawdopodobieństwo przeniesienia się, gdy klient odbył mało rozmów:", p_male)
print("Prawdopodobieństwo przeniesienia się, gdy klient odbył średnio rozmów:", p_srednie)
print("Prawdopodobieństwo przeniesienia się, gdy klient odbył dużo rozmów:", p_wysokie)
```

```python
#prawdopodobienstwo przeniesienia sie bez podzialu na liczbe rozmow
total = tabk.loc['All', 'All']
churn_true_count = tabk.loc[True, 'All']
```

```python
p_churn = churn_true_count / total
print("Prawdopodobieństwo przeniesienia bez podziału na rozmowy:", p_churn)
```

## Analiza tabeli krzyżowej
Tabela krzyżowa pokazuje rozkład liczby klientów, którzy odeszli (`True`) i pozostali (`False`) w zależności od kategorii rozmów (`CSC`).  

- Klienci z **małą liczbą rozmów** stanowią grupę referencyjną.  
- Klienci z **średnią liczbą rozmów** i **dużą liczbą rozmów** mają większą liczbę przypadków odejścia w stosunku do grupy referencyjnej.

## Wnioski

Na podstawie tabeli krzyżowej oraz dalszych obliczeń można zauważyć, że liczba rozmów
z biurem obsługi klienta ma związek z prawdopodobieństwem przeniesienia się klienta,
jednak zależność ta nie jest jednakowa dla wszystkich kategorii.

Iloraz szans dla klientów z małą liczbą rozmów wynosi 0.129, co oznacza, że w tej grupie
przeniesienia zdarzają się rzadko. Klienci, którzy rzadko kontaktują się z biurem obsługi,
zazwyczaj nie decydują się na zmianę operatora.

Dla klientów ze średnią liczbą rozmów iloraz szans wynosi 0.964, czyli jest bardzo zbliżony
do wartości 1. Oznacza to, że ryzyko przeniesienia w tej grupie jest praktycznie takie samo
jak w grupie klientów z małą liczbą rozmów. Różnice pomiędzy tymi dwiema grupami są bardzo
niewielkie i nie mają większego znaczenia praktycznego.

Z kolei w przypadku klientów z dużą liczbą rozmów iloraz szans wynosi 8.318. Jest to
bardzo wysoka wartość, która wskazuje, że klienci często kontaktujący się z biurem obsługi
klienta mają znacznie większą skłonność do przeniesienia się w porównaniu do klientów
z małą liczbą rozmów.

Współczynniki modelu logitowego potwierdzają te obserwacje. Wartość parametru b0 jest
ujemna, co oznacza niskie bazowe ryzyko przeniesienia dla klientów z małą liczbą rozmów.
Parametr b1, odpowiadający średniej liczbie rozmów, jest bardzo bliski zeru, co wskazuje,
że ta kategoria nie wpływa istotnie na zmianę ryzyka przeniesienia. Natomiast parametr b2
jest dodatni i wysoki, co potwierdza silny wpływ dużej liczby rozmów na wzrost ryzyka odejścia.

Obliczone prawdopodobieństwa przeniesienia pokazują, że około 11% klientów z małą
i średnią liczbą rozmów zdecydowało się na przeniesienie, natomiast w grupie klientów
z dużą liczbą rozmów odsetek ten przekracza 50%. Oznacza to, że istotny wzrost ryzyka
przeniesienia pojawia się dopiero przy dużej liczbie kontaktów z biurem obsługi klienta.

Dla porównania, prawdopodobieństwo przeniesienia obliczone bez podziału na liczbę rozmów
wynosi około 14.5%. Wartość ta jest wyższa niż w grupach z małą i średnią liczbą rozmów,
ale znacznie niższa niż w grupie klientów często kontaktujących się z obsługą.

Podsumowując, sama obecność kontaktów z biurem obsługi klienta nie musi oznaczać
zwiększonego ryzyka przeniesienia, jednak ich duża liczba jest wyraźnym sygnałem
niezadowolenia klienta i wiąże się z bardzo wysokim prawdopodobieństwem odejścia.
