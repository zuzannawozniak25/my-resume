```python
#importowanie bibliotek
import random
import matplotlib.pyplot as plt
import math 
```

```python
#zdefiniowanie atraktora

def w1_a(x, y):
    return 0.5*x - 0.5*y, 0.5*x + 0.5*y

def w2_a(x, y):
    return 0.5*x + 0.5*y + 0.5, -0.5*x + 0.5*y + 0.5

functions_a = [w1_a, w2_a]
```

```python
#zdefiniowanie n = 10000 punktów atraktora
n = 10000
```

```python
#wylosowanie osobno po jednej liczbie dla każdej współrzędnej z przedziału [0,1)
x = [random.random()]
y = [random.random()]
print("x =", x)
print("y =", y)
```

```python
#wykonanie pętli
for i in range(n - 1):
    r = random.random()
    
   #Podział przedziału na tyle rozłącznych podprzedziałów równej długości ile jest funkcji wk(x, y)
    if r < 0.5:
        xn, yn = functions_a[0](x[i], y[i])
    else:
        xn, yn = functions_a[1](x[i], y[i])
       
    x.append(xn)
    y.append(yn)
    
print(x[0], x[-1])
print(y[0], y[-1])
```

```python
#narysowanie wykresu
plt.scatter(x, y, s=0.1, color="deeppink")
plt.title("Atraktor (a)")
plt.show()
```

```python
#zdefiniowanie atraktora 

def w1_b(x, y):
    return 0.5*x, 0.5*y

def w2_b(x, y):
    return 0.5*(x + 1), 0.5*y

def w3_b(x, y):
    return 0.25*(2*x + 1), 0.25*(2*y + math.sqrt(3))

functions_b = [w1_b, w2_b, w3_b]
```

```python
#wylosowanie osobno po jednej liczbie dla każdej współrzędnej z przedziału [0,1)
x = [random.random()]
y = [random.random()]
print("x =", x)
print("y =", y)
```

```python
#wykonanie pętli
for i in range(n - 1):
    r = random.random()
    
   #Podział przedziału na tyle rozłącznych podprzedziałów równej długości ile jest funkcji wk(x, y)
    if r < 1/3:
        xn, yn = functions_b[0](x[i], y[i])
    elif 1/3 <= r < 2/3:
        xn, yn = functions_b[1](x[i], y[i])        
    else:
        xn, yn = functions_b[2](x[i], y[i])
       
    x.append(xn)
    y.append(yn)
    
print(x[0], x[-1])
print(y[0], y[-1])
```

```python
#narysowanie wykresu
plt.scatter(x, y, s=0.1, color="slateblue")
plt.title("Atraktor (b)")
plt.show()
```
