```python
import numpy as np 
import matplotlib.pyplot as plt
import sympy as sp
```

```python
#analitical solution

# defining symbols  
P, Q = sp.symbols('P Q', nonnegative=True)

# defining symbols  
r, K, s, u, a = sp.symbols('r K s u a', positive=True)

# model functions
Pn = P * (1 + r * (1 - P/K)) - s * P * Q
Qn = (1 - u) * Q + a * s * P * Q

# defining stationary equations P = Pn, Q = Qn
eq1 = sp.Eq(P, Pn)
eq2 = sp.Eq(Q, Qn)

# solutions
solutions = sp.solve((eq1, eq2), (P, Q), dict=True)
solutions
```

```python
#numerical simulation and visualisation
# defining parameters  
N = 100
P0 = 0.1
Q0 = 0.2
r = 0.5
u = 0.2
s = 2.2
a = 0.9
K = 0.7
```

```python
def Pn(P, Q):
    return P * (1 + r * (1 - P/K)) - s * P * Q

def Qn(P, Q):
    return (1 - u) * Q + a * s * P * Q
```

```python
P = np.zeros(N+1)
Q = np.zeros(N+1)
P[0] = P0
Q[0] = Q0

for n in range(N):
    P[n+1] = max(Pn(P[n], Q[n]), 0)
    Q[n+1] = max(Qn(P[n], Q[n]), 0)

# end results
print(f"P(100) = {P[-1]:.4f}")
print(f"Q(100) = {Q[-1]:.4f}")
```

```python
# time series plot
plt.figure(figsize=(8,4))
plt.plot(P, label="P (Rabbits)", marker='o', markersize=3)
plt.plot(Q, label="Q (Foxes)", marker='o', markersize=3)
plt.xlabel("Generations")
plt.ylabel("Foxes and Rabbits")
plt.title("P i Q Population Dynamics")
plt.grid(True)
plt.legend()
plt.show()

# phase portrait
plt.figure(figsize=(5,5))
plt.plot(P, Q, marker='o', markersize=3)
plt.xlabel("P (Rabbits)")
plt.ylabel("Q (Foxes)")
plt.title("Phase portrait")
plt.grid(True)
plt.show()
```
